# Exploring and Visualizing a Simple Dataset 

Objective:
 Learn how to load, inspect, and visualize a dataset to understand data trends and distributions.
Dataset:
 Iris Dataset (CSV format, can be loaded via seaborn or downloaded)

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Loading the dataset using pandas and seaborn. 
df=sns.load_dataset('iris')
```

```python
#Print the shape, column names, and the first few rows using .head(). 
df.shape
df.columns
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

# Visualizing data

```python
num_values=['sepal_length','sepal_width','petal_length','petal_width']
for col in num_values:
    plt.figure(figsize=(3,4))
    sns.histplot(df[col],kde=True,bins=20)
   
```

```python
df.head(1)
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(
    x="sepal_length",
    y="sepal_width",
    hue="species",
    data=df
)

plt.title("Iris Dataset Scatter Plot")

plt.show()
```

```python
plt.figure(figsize=(4,3))

sns.boxplot(
    x="species",
    y="petal_length",
    hue="species",
    data=df
)

plt.title("Petal Length Distribution")

plt.show()
```
